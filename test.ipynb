```python
import numpy as np
temp_results_pehe = []
temp_results_ate = []
data_name = "flickr_new_5"
for i in range(0, 10):
    temp_name = "dy_results/" + "test_results_" + data_name + "_DITE.DITE_split_" + str(i) + ".npy"
    temp = np.load(temp_name)
    temp_results_pehe.append(temp[0])
    temp_results_ate.append(temp[1])   
temp_results_pehe=np.array(temp_results_pehe)
temp_results_ate=np.array(temp_results_ate)
```

```python
print(np.mean(temp_results_pehe ** 0.5), np.std(temp_results_pehe ** 0.5))
print(np.mean(temp_results_ate), np.std(temp_results_ate))
```
